## Machine learning 

## For KNN

```python
import pandas as pd
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix
import joblib

# ----------------------------
# Load dataset
# ----------------------------
file_path = "ready_for_ml_class_1.csv"
df = pd.read_csv(file_path)

# Extract sample names, features, and labels
X = df.iloc[:, 1:-1]  # Features (excluding sample names and class labels)
y = df.iloc[:, -1]    # Class labels


# Split into training (80%) and testing (20%) sets while preserving class distribution
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=2673
)

# ----------------------------
# Define KNN hyperparameters for Grid Search
# ----------------------------
param_grid = {
    'n_neighbors': [1,3,5,7,9,11,13],          # Number of neighbors
    'weights': ['distance'] # Weighting function
}

# ----------------------------
# Grid Search with KNN
# ----------------------------
knn = KNeighborsClassifier()
grid_search = GridSearchCV(
    knn,
    param_grid,
    cv=5,
    scoring='accuracy',
    n_jobs=-1
)

grid_search.fit(X_train, y_train)

# ----------------------------
# Print mean CV accuracy for each hyperparameter combination
# ----------------------------
results = grid_search.cv_results_

print("\nMean CV Accuracy for each hyperparameter combination:\n")
for mean_score, params in zip(results['mean_test_score'], results['params']):
    print(f"Params: {params}, Mean CV Accuracy: {mean_score:.4f}")

# ----------------------------
# Best parameters
# ----------------------------
print("\nBest parameters:", grid_search.best_params_)
print("Best CV Accuracy:", grid_search.best_score_)

# ----------------------------
# Save the Best Model
# ----------------------------
best_model = grid_search.best_estimator_
joblib.dump(best_model, 'best_KNN_model.joblib')

# ----------------------------
# Test set evaluation
# ----------------------------
y_pred = best_model.predict(X_test)

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(cm)
```

## For Decision Tree

```python
import pandas as pd
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, confusion_matrix
import joblib

# ----------------------------
# Load dataset
# ----------------------------
file_path = "ready_for_ml_class_1.csv"
df = pd.read_csv(file_path)

# Extract features and labels
X = df.iloc[:, 1:-1]
y = df.iloc[:, -1]



# ----------------------------
# Train/test split
# ----------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=2673
)

# ----------------------------
# Hyperparameter grid
# ----------------------------
param_grid = {
    'max_depth': [50, 100, 200],
    'min_samples_split': [5, 10,20,50]
}

# ----------------------------
# Grid Search
# ----------------------------
dt = DecisionTreeClassifier(random_state=42)
grid_search = GridSearchCV(dt, param_grid, cv=5, scoring='accuracy', n_jobs=-1)
grid_search.fit(X_train, y_train)


# ----------------------------
# Print mean CV accuracy for each hyperparameter combination
# ----------------------------
results = grid_search.cv_results_

print("\nMean CV Accuracy for each hyperparameter combination:\n")
for mean_score, params in zip(results['mean_test_score'], results['params']):
    print(f"Params: {params}, Mean CV Accuracy: {mean_score:.4f}")

# ----------------------------
# Results
# ----------------------------
print("Best parameters:", grid_search.best_params_)
print("Best CV Accuracy:", grid_search.best_score_)

best_model = grid_search.best_estimator_
joblib.dump(best_model, 'best_DT_model.joblib')

y_pred = best_model.predict(X_test)
print("\nClassification Report:\n", classification_report(y_test, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

## For Random Forest

## For Gradient Boosting

```python
import pandas as pd
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import classification_report, confusion_matrix
import joblib

# ----------------------------
# Load dataset
# ----------------------------
file_path = "ready_for_ml_class_1.csv"
df = pd.read_csv(file_path)

# Extract features and labels
X = df.iloc[:, 1:-1]
y = df.iloc[:, -1]



# ----------------------------
# Train/test split
# ----------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=2673
)

# ----------------------------
# Hyperparameter grid
# ----------------------------
param_grid = {
    'n_estimators': [200,250,300],
    'max_depth': [1,2,3,4,5],
    'learning_rate': [0.01, 0.1]
}

# ----------------------------
# Grid Search
# ----------------------------
gb = GradientBoostingClassifier(random_state=42)
grid_search = GridSearchCV(gb, param_grid, cv=5, scoring='accuracy', n_jobs=-1)
grid_search.fit(X_train, y_train)


# ----------------------------
# Print mean CV accuracy for each hyperparameter combination
# ----------------------------
results = grid_search.cv_results_

print("\nMean CV Accuracy for each hyperparameter combination:\n")
for mean_score, params in zip(results['mean_test_score'], results['params']):
    print(f"Params: {params}, Mean CV Accuracy: {mean_score:.4f}")

# ----------------------------
# Results
# ----------------------------
print("Best parameters:", grid_search.best_params_)
print("Best CV Accuracy:", grid_search.best_score_)

best_model = grid_search.best_estimator_
joblib.dump(best_model, 'best_GB_model.joblib')

y_pred = best_model.predict(X_test)
print("\nClassification Report:\n", classification_report(y_test, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
```
